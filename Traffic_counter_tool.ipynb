# Building a counter

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import datetime
import csv
import os
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# 設定檔案儲存路徑與名稱
file_path = '/content/drive/My Drive/customer_log.csv'

# 初始化資料結構 (字典)
statistics = {
    '已進店': {'男': 0, '女': 0},
    '離開有購物': {'男': 0, '女': 0},
    '離開無購物': {'男': 0, '女': 0},
    '門外等候': {'男': 0, '女': 0}
}

# 建立按鈕
buttons = {}
situations = ['已進店', '離開有購物', '離開無購物', '門外等候']
genders = ['男', '女']

for situation in situations:
    buttons[situation] = {}
    for gender in genders:
        buttons[situation][gender] = widgets.Button(description=f"{gender}")

# 建立顯示區域
output = widgets.Output()
reset_button = widgets.Button(description="重設")
file_label = widgets.Label(value=f"檔案名稱: {file_path.split('/')[-1]}")
total_label = widgets.Label(value="總計: 0")

def update_total():
    total = 0
    for situation_data in statistics.values():
        total += sum(situation_data.values())
    total_label.value = f"總計: {total}"

def save_data_to_drive(situation, gender):
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open(file_path, 'a', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        writer.writerow([timestamp, situation, gender])

def handle_button_click(button, situation, gender):
  global statistics
  statistics[situation][gender] += 1
  save_data_to_drive(situation, gender)
  update_total()

def reset_statistics(button):
    global statistics
    statistics = {
        '已進店': {'男': 0, '女': 0},
        '離開有購物': {'男': 0, '女': 0},
        '離開無購物': {'男': 0, '女': 0},
        '門外等候': {'男': 0, '女': 0}
    }
    # 刪除Google Drive 的數據
    if os.path.exists(file_path):
        os.remove(file_path)

    update_total()
    # 初始化 CSV 檔案的表頭
    with open(file_path, 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['時間', '情況', '性別'])

    with output:
        clear_output()
        print("數據已重設")

# 建立按鈕的事件處理
for situation in situations:
    for gender in genders:
        button = buttons[situation][gender]
        button.on_click(lambda b, situation=situation, gender=gender: handle_button_click(b, situation, gender))
#設定重設按鈕的事件處理
reset_button.on_click(reset_statistics)
# 建立介面佈局
layout = widgets.Layout(width='auto', height='auto')
situation_widgets = []
for situation in situations:
  gender_buttons = [buttons[situation][gender] for gender in genders]
  situation_layout = widgets.HBox(gender_buttons, layout=layout)
  situation_label = widgets.Label(value=situation)
  situation_vbox = widgets.VBox([situation_label, situation_layout], layout=layout)
  situation_widgets.append(situation_vbox)

main_layout = widgets.VBox(situation_widgets + [reset_button, file_label, total_label, output], layout=layout)

# 顯示介面
display(main_layout)

# 初始化 CSV 檔案的表頭
if not os.path.exists(file_path):
    with open(file_path, 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['時間', '情況', '性別'])
```
